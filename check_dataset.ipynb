```python
import pandas as pd

df = pd.read_csv("data.csv")
df.head()
```

```python
print(f"Total number of items (folders): {len(df)}")
```

```python
assert df["id"].is_unique, "Duplicate IDs found!"
print("All IDs are unique:", df["id"].is_unique)
```

```python
print("Missing values per column:")
print(df.isna().sum())
```

```python
print("Folders by start_year:")
print(df["start_year"].value_counts().sort_index())

print("\nAuthored folders vs. background source dossiers:")
print(df["is_source_dossier"].value_counts())

print("\nTop 10 primary topics:")
topics = df["primary_topics"].str.split("; ").explode()
print(topics.value_counts().head(10))
```
